```python
import subprocess
import py_compile
import re
from pathlib import Path

print("=" * 80)
print("🔧 ВОССТАНОВЛЕНИЕ КОНФИГУРАЦИИ ЯДРА (core_config.py)")
print("=" * 80)

root = Path(".").resolve()

# 1. Достаем оригинальный файл из коммита Git
subprocess.run(["git", "checkout", "HEAD", "--", "bot_config.py"], check=True)
print("✔ Исходный bot_config.py успешно восстановлен из Git.")

raw_code = (root / "bot_config.py").read_text(encoding="utf-8")

# 2. Вычищаем строки Telegram
clean_lines = []
for line in raw_code.splitlines():
    if any(k in line for k in ["TELEGRAM_BOT_TOKEN", "TELEGRAM_CHAT_ID", "# Telegram"]):
        continue
    clean_lines.append(line)

core_config_content = "\n".join(clean_lines) + "\n"

# 3. Сохраняем в core_config.py
core_cfg_path = root / "core_config.py"
core_cfg_path.write_text(core_config_content, encoding="utf-8")
py_compile.compile(str(core_cfg_path), doraise=True)
print("✔ Создан чистый канонический core_config.py (без параметров Telegram).")

# 4. Настраиваем bot_config.py как backward-compatibility прокси
proxy_code = '''"""
Backward compatibility proxy for Core Configuration.
Канонический конфиг ядра находится в core_config.py.
"""
from core_config import *
'''
bot_cfg_path = root / "bot_config.py"
bot_cfg_path.write_text(proxy_code, encoding="utf-8")
py_compile.compile(str(bot_cfg_path), doraise=True)
print("✔ bot_config.py переведен в режим прозрачного прокси на core_config.py.")

print("=" * 80)
```
