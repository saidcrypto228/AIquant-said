```python
import json
import re
from pathlib import Path

print("=" * 75)
print("ШАГ 2: КАРТИРОВАНИЕ ВХОЖДЕНИЙ (DRY RUN БЕЗ ЗАПИСИ)")
print("=" * 75)

TARGET_EXTS = {
    ".py",
    ".json",
    ".md",
    ".html",
    ".txt",
    ".env",
    ".sh",
    ".service",
    ".conf",
    ".ipynb",
}
IGNORE_DIRS = {
    "__pycache__",
    ".git",
    ".ipynb_checkpoints",
    "backups",
    "_qvex_safety_backups",
}

findings = {"brand_matches": [], "term_matches": []}

brand_pattern = re.compile(
    r"(qvex-said|qvex|a-iquant)", re.IGNORECASE
)  # Целевые бренды
generic_quant_pattern = re.compile(
    r"\bquant\b", re.IGNORECASE
)  # Отдельное слово 'quant'

for fpath in Path(".").rglob("*"):
    if not fpath.is_file() or fpath.suffix not in TARGET_EXTS:
        continue
    if any(part in IGNORE_DIRS for part in fpath.parts):
        continue

    try:
        text = fpath.read_text(encoding="utf-8", errors="ignore")
    except Exception:
        continue

    # Поиск бренда (QVEX / qvex)
    b_matches = list(brand_pattern.finditer(text))
    if b_matches:
        findings["brand_matches"].append((fpath, len(b_matches)))

    # Поиск отдельного слова quant
    q_matches = list(generic_quant_pattern.finditer(text))
    if q_matches:
        findings["term_matches"].append((fpath, len(q_matches)))

print(f"1. Файлы с прямым именем проекта (QVEX / qvex) — ПОДЛЕЖАТ ЗАМЕНЕ:")
for p, count in findings["brand_matches"]:
    print(f"   • {str(p):<45} (вхождений: {count})")

print(f"\n2. Файлы со словом 'quant' (требуют разделения: бренд или математика):")
for p, count in findings["term_matches"][:15]:
    print(f"   • {str(p):<45} (вхождений: {count})")
if len(findings["term_matches"]) > 15:
    print(f"   ... и еще {len(findings['term_matches']) - 15} файлов")

print("\n" + "=" * 75)
print("Аудит завершен. Данные на диске не изменялись.")
print("=" * 75)
```
